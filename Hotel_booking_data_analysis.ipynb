```python
import calendar

data['arrival_month'] = data['arrival_month'].apply(
    lambda x: calendar.month_name[x]
)
```

```python
data['arrival_month'].unique()
```

```python
dict_month = {'October':10, 'November':11, 'February':2, 'May':4, 'April':3, 'September':9,
       'December':12, 'July':7, 'June':6, 'August':8, 'March':3, 'January':1}
```

```python
data['arrival_month_index'] = data['arrival_month'].map(dict_month)
```

```python
data[['arrival_year','arrival_month_index','arrival_month']]
```

```python
data['arrival_date'] = pd.to_datetime(
    data['arrival_year'].astype(str) + '-' +
    data['arrival_month_index'].astype(str).str.zfill(2) + '-01'
)

data['arrival_date'] = data['arrival_date'].dt.strftime('%Y-%m-%d')

data[['arrival_year', 'arrival_month_index', 'arrival_month', 'arrival_date']].head()
```
